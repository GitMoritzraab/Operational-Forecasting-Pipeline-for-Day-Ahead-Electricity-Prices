# SQRA interactive inspection

The canonical SQRA implementation lives in `sqra_model.py` and `run_sqra.py`. This notebook is optional and never used by `run_full_experiment.py`.

```python
from pathlib import Path
import sys

start = Path.cwd().resolve()
REPO_ROOT = next(
    candidate for candidate in (start, *start.parents)
    if (candidate / 'experiment_config.py').is_file()
)
sys.path.insert(0, str(REPO_ROOT))

from experiment_config import load_experiment_config
from experiment_manifest import SQRA_RUNS
from pipeline.delivery_index import read_forecast_csv
from pipeline.sqra.run_sqra import run_sqra_experiment

EXPERIMENT = load_experiment_config(REPO_ROOT)
CONFIG_NAME = 'era5_fundamental'
SQRA_RUNS[CONFIG_NAME]
```

## Optional execution

Change the guard deliberately before running this cell. It writes the same four SQRA output files as the command-line entry point.

```python
RUN_EXPERIMENT = False
if RUN_EXPERIMENT:
    artifacts = run_sqra_experiment(CONFIG_NAME, EXPERIMENT)
    display(artifacts.forecast.head())
    display(artifacts.metrics)
```

## Inspect an existing result

```python
import json
import pandas as pd

result_dir = SQRA_RUNS[CONFIG_NAME].output_dir(EXPERIMENT.results_root)
if (result_dir / 'forecast.csv').exists():
    forecast = read_forecast_csv(result_dir / 'forecast.csv', require_complete_days=True)
    metrics = pd.read_csv(result_dir / 'metrics.csv')
    metadata = json.loads((result_dir / 'config.json').read_text(encoding='utf-8'))
    display(metadata)
    display(forecast.head())
    display(metrics)
else:
    print(f'No existing result at {result_dir}')
```
