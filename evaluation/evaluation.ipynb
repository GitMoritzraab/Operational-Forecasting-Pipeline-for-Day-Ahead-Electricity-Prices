# Evaluation interactive inspection

The canonical calculations live in `evaluation_core.py`; `run_evaluation.py` is the automated entry point. This notebook is optional and contains no independent statistical implementation.

```python
from pathlib import Path
import sys

start = Path.cwd().resolve()
REPO_ROOT = next(
    candidate for candidate in (start, *start.parents)
    if (candidate / 'experiment_config.py').is_file()
)
sys.path.insert(0, str(REPO_ROOT))

from experiment_config import load_experiment_config
from evaluation.evaluation_core import LABEL_MAP_SQRA, plot_gw_heatmap
from evaluation.run_evaluation import run_evaluation

EXPERIMENT = load_experiment_config(REPO_ROOT)
EXPERIMENT
```

## Optional execution

This reads all 34 forecast files, prints the numerical tables, and rewrites the four GW PDFs. Change the guard deliberately before running.

```python
RUN_EVALUATION = False
if RUN_EVALUATION:
    results = run_evaluation(EXPERIMENT)
```

## Inspect returned tables and figures

```python
if 'results' in globals():
    display(results.point_metrics)
    display(results.median_mae)
    display(results.aps_summary)
    for name, table in results.coverage.items():
        print(name)
        display(table)
    display(plot_gw_heatmap(
        results.aps_gw, figsize=(6, 4.5), fontsize=13,
        label_map=LABEL_MAP_SQRA, fraction=0.03, pad=0.01,
    ))
else:
    print('Set RUN_EVALUATION=True and run the previous cell first.')
```
